# Simandou face aux autres grandes mines de fer

**Question :** le scénario de Simandou (étape 4) est-il réaliste ? Et que pèsera la Guinée dans le fer mondial ?

**Pourquoi pas de machine learning ici :** Simandou a démarré fin 2025. Il n'y a pas d'historique, donc rien à apprendre pour un modèle. On fait ce que font les analystes dans ce cas : on **compare** avec des mines semblables qui ont déjà fait leur montée en charge (**prévision par analogie**).

**Les deux mines de comparaison :**
- **Roy Hill** (Australie) : 55 Mt/an, nouvelle mine avec sa propre voie ferrée (344 km) et son port. Première expédition en décembre 2015.
- **S11D** (Brésil, Vale) : 90 Mt/an, la plus grande mine de fer jamais construite par Vale. Démarrage fin 2016.

**Données :**
- `data/montee_en_charge_mines.csv` : production réelle année par année, avec le lien vers chaque rapport.
- `data/fer_monde.csv` : production mondiale de minerai de fer en 2025 par pays (USGS).
- `data/simandou.csv` et `data/hypotheses_simandou.csv` : faits et scénario de l'étape 4.

## 1. Les montées en charge comparées

```python
montee = requete("06_montee_en_charge").df()
display(montee)
graphique_montee_en_charge(montee)
```

**Constat :** on met les trois mines sur la même échelle, en **% de leur capacité annoncée**, l'année 1 étant la première année après le démarrage.
- **Roy Hill :** 60 % dès l'année 1, 100 % à l'année 3. Rapide, mais en retard sur son annonce : en mars 2016, Roy Hill promettait 55 Mt/an fin 2016 ; le rythme a été atteint en septembre 2017, 9 mois plus tard.
- **S11D :** 25 %, puis 64 %, puis 82 %. Son meilleur score est de 92 % (année 4), et elle **n'a jamais atteint 100 %**.
- **Simandou (scénario) :** 6 % en 2026, 21 % en 2027. C'est **bien plus lent** que les deux autres au début. Ce n'est pas un choix au hasard : c'est la prévision de ventes de Rio Tinto pour 2026 (5 à 10 Mt), et le rythme réel de SimFer au 3e trimestre 2026 était d'environ 10 Mt/an pour une capacité de 60 Mt.

Pourquoi plus lent ? Simandou cumule tout : une mine dans une zone isolée, plus de 600 km de voie ferrée neuve, un port neuf, et deux opérateurs différents (SimFer et WCS) qui doivent se coordonner.

## 2. Et si Simandou suivait le rythme de Roy Hill ou de S11D ?

**Constat :** si Simandou montait en charge comme Roy Hill, il produirait déjà 72 Mt en 2026. Comme S11D, 30 Mt. La réalité de 2026 (5 à 10 Mt) montre que **les deux analogies sont trop rapides pour le début**. Simandou a déjà au moins un an de retard sur ces rythmes.

## 3. Le plein régime de 120 Mt est-il réaliste ?

```python
requete("06_niveau_atteint").df()
```

**Constat :** une fois la montée en charge terminée (à partir de l'année 3), S11D tourne en moyenne à **83 %** de sa capacité, et au mieux à 92 %. Roy Hill a atteint 100 %, mais on n'a qu'une année de données ici.

Le scénario de l'étape 4 suppose que Simandou atteint **100 %** dès 2030. Si Simandou faisait comme S11D (83 %), il produirait environ **100 Mt** au lieu de 120 Mt. Au prix moyen de 97 $/t, cela représente environ **9,7 milliards de dollars** d'exportations au lieu de 11,6, soit environ 2 milliards de moins par an.

## 4. Que pèsera la Guinée dans le fer mondial ?

```python
classement = requete("06_fer_classement").df()
display(classement.head(10))
graphique_fer_classement(classement)
```

**Constat :** en 2025, le monde a produit environ **2,6 milliards de tonnes** de minerai de fer. L'Australie en produit à elle seule 38 %.

Avec 120 Mt, la Guinée deviendrait le **5e producteur mondial**, derrière l'Australie, le Brésil, l'Inde et la Chine, et devant l'Iran et la Russie. Elle pèserait environ **4,6 %** de la production mondiale. Même à 100 Mt, elle resterait 5e.

C'est beaucoup moins que pour la bauxite, où la Guinée fait 34 % du monde (notebook 05). Sur le fer, la Guinée sera un acteur important, mais pas dominant. Le minerai de Simandou a toutefois un atout : sa **teneur de 65 % de fer**, plus élevée que la plupart des minerais australiens, ce qui le rend recherché par les aciéries chinoises.

## Ce qu'il faut retenir

1. **Sans historique, pas de machine learning :** on compare avec des mines semblables (prévision par analogie).
2. **Simandou démarre plus lentement** que Roy Hill et S11D : 6 % de sa capacité la première année, contre 60 % et 25 %.
3. **Le plein régime n'est pas garanti :** S11D n'a jamais atteint 100 % et tourne autour de 83 %. Un Simandou à 100 Mt au lieu de 120 Mt, c'est environ 2 milliards de dollars d'exportations en moins par an.
4. **5e producteur mondial de fer** avec Simandou à plein régime, mais environ 5 % du marché seulement, contre 34 % pour la bauxite.

## Limites

- **Deux mines de comparaison seulement**, et aucune n'est exactement comme Simandou (taille, pays, infrastructures).
- **Roy Hill est mesuré en années fiscales** (juillet à juin) et en ventes. S11D est mesuré en années civiles et en production.
- **Le classement mondial est celui de 2025.** D'ici 2030, les autres pays auront aussi changé.
